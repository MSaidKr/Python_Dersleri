```python
#Exercise 1

list1 = [20,13,18]
def item(row,i):
  return row[i]

a = item(list1,2)
print(a)

#Exercise 2

def push(row,x):
  row.append(x)
  return row

push(list1,'a')
print(list1)

#Exercise 3

def pull(row,x):
  row.remove(x)
  return row

pull(list1,'a')
print(list1)

pull(list1,18)
print(list1)

#Exercise 4

def pair(a,b):
  return (a,b)

b = pair(20,13)
print(b)

#Exercise 5

D = {'a':1,'b':2,'c':5}
def value(d,key):
  return d[key]

y = value(D,'a')
print(y)

#Exercise 6

def put(d,key,value):
  d[key] = value
  return d

print(D)
put(D,'d',4)
print(D)
put(D,'c',3)
print(D)

#Exercise 7

def drop(d,key):
  del d[key]
  return d

print(D)
drop(D,'d')
print(D)

#Exercise 8

def get(d,key):
  return d.get(key)

a = get(D,'a')
print(a)
f = get(D,'f')
print(f)

def get2(d,key):
  if key in d:
    return d[key]
  else:
    return None

c = get2(D,'c')
print(c)
e = get2(D,'e')
print(e)

#Task 1:

liste = ['bir','isim','listesi','yapalım']
initials_list = []
def initials(a):
  for i in a:
    initials_list.append(i[0])
  return initials_list

initials(liste)
print(initials_list)

#Task 2:

def vowels(word):
  sayac = 0
  for i in word:
    if i in 'aeiouAEIOU':
      sayac += 1
  return sayac

vowels('capasitor')

#Task 3:

def more_vowels(a):
  sayac1 = 0
  sayac2 = 0
  for i in a:
    if i in 'aeiouAEIOU':
      sayac1 += 1
    else:
      sayac2 +=1
  if sayac1 > sayac2:
    return True
  else:
    return False

more_vowels('capasitor')

#Task 4:

listE = []
def span(xs):
    if not xs:
        return None

    kucuk = xs[0]
    buyuk = xs[0]

    for x in xs:
        if x < kucuk:
            kucuk = x
        if x > buyuk:
            buyuk = x

    return (kucuk, buyuk)

t = span(list1)
r = span(listE)
print(t)
print(r)

#Task 5:

listx = [1,2,3,4,5,6,7,8,8,8,9]
def times(xs,x):
  sayac = 0
  for i in xs:
    if i == x:
      sayac += 1
    else:
      sayac += 0
  return sayac

f = times(listx,8)
print(f)

# task 6:

def tally(text):
  word = {}
  for i in text.lower():
    if i.isalpha():
      if i in word:
        word[i] += 1
      else:
        word[i] = 1
    else:
      pass
  return word

z = tally('Ca1aSiToRr:!')
print(z)

#Task 7:

listString = ['gercekten','cok','zevksiz','bir','zaman','geciriyorum']
def group(words):
  key = []
  d = {}
  for i in words:
    if i[0] not in key:
      key.append(i[0])
      d[i[0]] = []
    else:
      pass
  for i in words:
    if i[0] in key:
      d[i[0]].append(i)
    else:
      pass
  return (d,key)

y = group(listString)
print(y)

#Task 8:

def pile(words):
  if words == []:
    return None
  else:
    d = {}
    ss = []
    for i in words:
      sesli = 0
      for s in i:
        if s in 'AEIOUaeiou':
          sesli += 1
      if sesli not in d:
        d[sesli] = []
        ss.append(sesli)
      d[sesli].append(i)
  max = -1
  en = None
  for f in ss:
    grup = d[f]
    kelime_sayisi = 0
    for kelime in grup:
      kelime_sayisi += 1
    if kelime_sayisi > max:
      max = kelime_sayisi
      en = f
  return (en, d[en])
s = pile(listString)
print(s)
```
