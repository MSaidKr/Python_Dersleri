# **Connect 4**

```python
def create_board():
    board = []
    for i in range(6):
        x = ['.','.','.','.','.','.','.']
        board.append(x)
    return board

def print_board(board):
    for row in board:
        print(" ".join(row))
    print("1 2 3 4 5 6 7")

def drop_piece(board,col,player):
    place = False
    for i in range(5,-1,-1):
        if board[i][col] == 'X' or board[i][col] == 'O':
            pass
        else:
            board[i][col] = player
            place = True
            break
    return place

#Her çapraz doğrultusu için 6 tane, dikey için 1 ve yatay için 1 tane doğrulama.
def check_win(board,player):
    w = False
    sayac = 0
    #yatayda doğrulama
    for i in range(6):
        sayac = 0
        for j in range(7):
            if board[i][j] == player:
                sayac +=1
                if sayac == 4:
                    w = True
            else:
                sayac = 0
    sayac = 0
    #dikeyde doğrulama
    for i in range(7):
        sayac = 0
        for j in range(6):
            if board[j][i] == player:
                sayac += 1
                if sayac == 4:
                    w = True
            else:
                sayac = 0
    sayac = 0
    #aşağı 1 numara
    for i in range(4):
        if board[i + 2][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #aşağı 2 numara
    for i in range(5):
        if board[i + 1][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #aşağı 3 numara
    for i in range(6):
        if board[i][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #aşağı 4 numara
    for i in range(6):
        if board[i][i + 1] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #aşağı 5 numara
    for i in range(5):
        if board[i][i + 2] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #aşağı 6 numara
    for i in range(4):
        if board[i][i + 3] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 1 numara
    for i in range(4):
        j = 3-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 2 numara
    for i in range(5):
        j = 4-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 3 numara
    for i in range(6):
        j = 5-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 4 numara
    for i in range(1,7):
        j = 6-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 5 numara
    for i in range(2,7):
        j = 7-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    #yukarı 6 numara
    for i in range(3,7):
        j = 8-i
        if board[j][i] == player:
            sayac += 1
            if sayac == 4:
                w = True
        else:
            sayac = 0
    sayac = 0
    return w

def main():
    w = False
    b = create_board()
    print('The game is begining. Choose someone to star(you can flip a coin). The first player is X. Whenever you want to quit, you can write (çık).')
    print_board(b)
    for i in range(21):
        a = i + 1
        print(f'Round: {a}')
        p = 'X'
        while True:
            c = input('First player putting an X on, column number...')
            if c == 'çık':
                print('First player has left the game. The winner is second player.')
                return
            if not c.isdigit():
                print('Unvalid number detected. Please try again.')
                continue
            c = int(c)
            if c < 1 or c > 7:
                print('Unvalid column number. Please try again.')
                continue
            if drop_piece(b,c-1,p):
                break
            else:
                print('This column is full.')
        print_board(b)
        w = check_win(b,p)
        if w == True:
            print('The game over. The winner: First player (X)')
            return
        p = 'O'
        while True:
            c = input('Second player putting an O on, column number...')
            if c == 'çık':
                print('Second player has left the game. The winner is first player.')
                return
            if not c.isdigit():
                print('Unvalid number detected. Please try again.')
                continue
            c = int(c)
            if c < 1 or c > 7:
                print('Unvalid column number. Please try again.')
                continue
            if drop_piece(b,c-1,p):
                break
            else:
                print('This column is full.')
        print_board(b)
        w = check_win(b,p)
        if w == True:
            print('The game over. The winner: Second player (O)')
            return
    print_board(b)
    print(' ')
    print("The game is over. It's a draw!")


main()
```
