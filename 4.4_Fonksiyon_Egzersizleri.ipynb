```python
#Task 1
def correct_func(n):
    a = []
    b = []
    sayac = 0
    for i in range(n+1):
        a.append(i)
    for j in a:
        j = str(j)
        h = list(j)
        for t in h:
            b.append(t)
    for k in b:
        if k == '1':
            sayac +=1
    return sayac

b =correct_func(11)
print(b)


def incorrect_func(n):
    sayı=0
    for i in range(n+1):
        stri = str(i)
        if '1' in stri:
            sayı+=1
    return sayı


v = incorrect_func(11)
print(v)

def task1(ffunc,n):
    a = []
    b = []
    sayac = 0
    for i in range(n+1):
        a.append(i)
    for j in a:
        j = str(j)
        h = list(j)
        for t in h:
            b.append(t)
    for k in b:
        if k == '1':
            sayac +=1
    if sayac == ffunc(n):
        return True
    else:
        return False

z = task1(correct_func,11)
print(z)

#Task 2

def task2(s):
    s = s.lower()
    sayac = 0
    liste = []
    ikilikler = []
    double = ''
    s = list(s)
    for i in s:
        liste.append(i)
        sayac += 1
    for i in range(sayac):
        for j in range(sayac):
            if i != j:
                double = ''.join(liste[i] + liste[j])
                if double not in ikilikler:
                    ikilikler.append(double)
    for tek in liste:
        if tek not in ikilikler:
            ikilikler.append(tek)
    ikilikler.sort()
    return ikilikler

t2 = task2('karaka')
print(t2)

#Task 3

def task3(s):
    sayı = 0
    sayac = 1
    harf =[]
    çıktı = ''
    s = list(s)
    for i in s:
        harf.append(i)
        sayı += 1
    if sayı == 0:
        return('',0)
    for i in range(1, sayı):
        if harf[i] == harf[i-1]:
            sayac += 1
        else:
            çıktı += harf[i-1]
            if sayac > 1:
                sayac = str(sayac)
                çıktı += sayac
            sayac = 1
    çıktı += harf[-1]
    if sayac > 1:
        çıktı += str(sayac)
    oran = round(((sayı - len(çıktı)) / sayı) * 100)
    return (çıktı, oran)

y3 = task3('dddssssaaaaAADAF')
print(y3)
```
