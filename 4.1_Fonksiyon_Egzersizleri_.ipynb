```python
#Exercise 1
def fibonacci(n):
    if n == 1 or n==2:
        return 1
    return fibonacci(n-1) + fibonacci(n -2)
a = fibonacci(6)
print(a)
b = fibonacci(4)
print(b)

#Exercise 2
def factorial(n):
    n = int(n)
    if n == 0:
        return 1
    return factorial(n - 1) * n
c = factorial(4)
print(c)

#Task 1
# i = row / j = column
def func1(row,column):
    row = int(row)
    column = int(column)
    if row == 1 and column == 1:
        return 1
    if column == 1:
        return 3
    if column == row:
        return 2
    return func1(row - 1, column - 1) + func1(row - 1, column)
d = func1(9,1)
print(d)
e = func1(4,3)
print(e)

#Task 2
def func2(s1,s2):
    s1 = list(s1)
    l1 = len(s1)
    s2 = list(s2)
    l2 = len(s2)
    sayac = 0
    if l1<l2:
        for i in range(l1):
            if s1[i] == s2[i]:
                sayac += 1
    if l1>=l2:
        for i in range(l2):
            if s2[i] == s1[i]:
                sayac += 1
    return sayac
#ikinci daha uzun
f = func2('para','parlak')
print(f)
#ortak harf yok
g = func2('kıdem','çanta')
print(g)
#ilki daha uzun
h = func2('kabataş','kap')
print(h)

#Task 3
def contains_digit(string):
    for i in string:
        if i.isdigit():
            return True
    return False

def check_username(un):
    if len(un) < 5:
        return 'Too short'
    if not contains_digit(un):
        return 'Digit required'
    return 'Valid'

un1 = check_username('ay')
print(un1)
un2 = check_username('kelebek')
print(un2)
un3 = check_username('kelebek41')
print(un3)
```
